```python
import torch.nn as nn
import torch
```

```python
model = nn.Transformer(
    d_model=512,
    nhead=8,
    num_encoder_layers=6,
    num_decoder_layers=6,
    dim_feedforward=2048,
    dropout=0.1,
    activation="relu",
    batch_first=True
)
```

```python
print(model)
```

```python
src = torch.rand(10, 32, 512)
tgt = torch.rand(10, 42, 512)
out = model(src, tgt)
print(out.shape)
```
